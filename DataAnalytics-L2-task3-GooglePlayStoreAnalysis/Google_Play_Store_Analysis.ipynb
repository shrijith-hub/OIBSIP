# Cleaning the Size Column

The `Size` column contains values in different units:

- Megabytes (M)
- Kilobytes (k)
- "Varies with device"

We convert all sizes to megabytes (MB) for consistent analysis.

```python
def convert_size(size):
    if size == "Varies with device":
        return np.nan

    if size.endswith("M"):
        return float(size[:-1])

    if size.endswith("k"):
        return float(size[:-1]) / 1024

    return np.nan
```

```python
apps["Size"] = apps["Size"].apply(convert_size)
```

# Revenue Estimation

The Google Play Store dataset does not contain actual revenue information.

An estimated revenue is calculated using:

**Estimated Revenue = Price × Installs**

This provides an approximate measure of the potential revenue generated by paid applications.

```python
apps["Estimated Revenue"] = apps["Price"] * apps["Installs"]
```

```python
top_revenue = (
    apps.sort_values(
        by="Estimated Revenue",
        ascending=False
    )
    [["App", "Category", "Price", "Installs", "Estimated Revenue"]]
    .head(10)
)

top_revenue
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=top_revenue,
    x="Estimated Revenue",
    y="App",
    palette="rocket"
)

plt.title("Top 10 Apps by Estimated Revenue")
plt.xlabel("Estimated Revenue")
plt.ylabel("Application")

plt.show()
```

### Insight

- A small number of paid apps contribute disproportionately to estimated revenue.
- Premium pricing combined with high install counts results in the greatest revenue potential.
- Most applications generate little estimated revenue because they are free.

```python
reviews = reviews.dropna(subset=["Translated_Review"]).reset_index(drop=True)
```

```python
reviews.isnull().sum()
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()
```

```python
reviews["Sentiment Score"] = reviews["Translated_Review"].apply(
    lambda text: sia.polarity_scores(str(text))["compound"]
)
```

```python
def classify_sentiment(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

reviews["Predicted Sentiment"] = reviews["Sentiment Score"].apply(classify_sentiment)
```

```python
reviews.head()
```

```python
reviews["Predicted Sentiment"].value_counts()
```

# Sentiment Distribution
