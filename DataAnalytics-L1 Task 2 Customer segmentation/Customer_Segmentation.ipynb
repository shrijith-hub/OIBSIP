# Customer Analysis

In this section, we analyze customer purchasing behavior by calculating total sales, total profit, and the number of orders placed by each customer.

```python
df["Customer Name"].nunique()
```

### Observation

The dataset contains multiple unique customers, allowing meaningful customer segmentation based on purchasing behavior.

```python
customer_summary = df.groupby("Customer Name").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Order ID": "count"
})
```

```python
df.groupby("Customer Name")
```

```python
customer_summary.rename(columns={"Order ID": "Frequency"}, inplace=True)
```

```python
customer_summary.head()
```

```python
top_sales = customer_summary.sort_values(by="Sales", ascending=False)

top_sales.head(10)
```

### Observation

The table displays the top 10 customers based on total sales. These customers contribute significantly to the company's revenue and may be valuable for customer retention strategies.

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=top_sales.head(10),
    x="Sales",
    y=top_sales.head(10).index
)

plt.title("Top 10 Customers by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Customer Name")

plt.show()
```

### Observation

The bar chart clearly shows the customers generating the highest sales. Retaining these high-value customers can significantly impact overall business revenue.

```python
top_profit = customer_summary.sort_values(by="Profit", ascending=False)

top_profit.head(10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=top_profit.head(10),
    x="Profit",
    y=top_profit.head(10).index
)

plt.title("Top 10 Customers by Profit")
plt.xlabel("Total Profit")
plt.ylabel("Customer Name")

plt.show()
```

# Customer Segmentation

Customer segmentation is the process of dividing customers into different groups based on their purchasing behavior.

In this project, customers are segmented based on their total sales into:
- High Value Customers
- Medium Value Customers
- Low Value Customers

```python
customer_summary["Sales"].describe()
```

```python
def segment_customer(sales):
    if sales >= 5000:
        return "High Value"
    elif sales >= 2000:
        return "Medium Value"
    else:
        return "Low Value"
```

```python
customer_summary["Segment"] = customer_summary["Sales"].apply(segment_customer)
```

```python
customer_summary.head()
```

```python
customer_summary["Segment"].value_counts()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=customer_summary,
    x="Segment",
    order=["High Value", "Medium Value", "Low Value"]
)

plt.title("Customer Segments")
plt.xlabel("Segment")
plt.ylabel("Number of Customers")

plt.show()
```

### Observation

Most customers belong to the Low Value segment, while only a smaller percentage fall into the High Value segment. This indicates an opportunity for the business to increase customer value through targeted marketing and loyalty programs.

# Average Sales by Customer Segment

This analysis compares the average sales generated by customers in each segment.

```python
customer_summary.groupby("Segment")["Sales"].mean()
```

```python
customer_summary.groupby("Segment")["Profit"].mean()
```

```python
customer_summary.groupby("Segment")["Frequency"].mean()
```
