```python
df = pd.read_csv(
    "super store dataset/Sample - Superstore.csv",
    encoding="latin1"
)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
subcategory_sales = (
    df.groupby('Sub-Category')['Sales']
      .sum()
      .sort_values(ascending=False)
)

subcategory_sales
```

```python
plt.figure(figsize=(14, 6))

ax = sns.barplot(
    x=subcategory_sales.index,
    y=subcategory_sales.values
)

plt.title("Total Sales by Sub-Category", fontsize=14, fontweight='bold')
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

# Add value labels
for i, value in enumerate(subcategory_sales.values):
    ax.text(i, value + 5000, f"{value:,.0f}",
            ha='center', fontsize=8)

plt.show()
```

```python
subcategory_profit = (
    df.groupby('Sub-Category')['Profit']
      .sum()
      .sort_values(ascending=False)
)

subcategory_profit
```

```python
plt.figure(figsize=(14, 6))

ax = sns.barplot(
    x=subcategory_profit.index,
    y=subcategory_profit.values
)

plt.title("Total Profit by Sub-Category", fontsize=14, fontweight='bold')
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)

# Add value labels
for i, value in enumerate(subcategory_profit.values):
    ax.text(i, value + 1000, f"{value:,.0f}",
            ha='center', fontsize=8)

plt.show()
```

```python
region_sales = (
    df.groupby('Region')['Sales']
      .sum()
      .sort_values(ascending=False)
)

region_sales
```

```python
plt.figure(figsize=(8,5))

ax = sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

plt.title("Total Sales by Region", fontsize=14, fontweight='bold')
plt.xlabel("Region")
plt.ylabel("Total Sales")

# Add value labels
for i, value in enumerate(region_sales.values):
    ax.text(i, value + 5000, f"{value:,.0f}",
            ha='center', fontsize=9)

plt.show()
```

```python
# Create a Month-Year column
df['Month-Year'] = df['Order Date'].dt.to_period('M')
```

```python
monthly_sales = (
    df.groupby('Month-Year')['Sales']
      .sum()
)

monthly_sales
```

```python
plt.figure(figsize=(14,6))

monthly_sales.plot(marker='o')

plt.title("Monthly Sales Trend", fontsize=14, fontweight='bold')
plt.xlabel("Month-Year")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

## Observation

- Sales fluctuate across different months.
- Some months show significant peaks, indicating periods of high demand.
- The company may investigate these high-performing months to understand the factors contributing to increased sales.

```python
top_customers = (
    df.groupby('Customer Name')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```

```python
plt.figure(figsize=(12,6))

ax = sns.barplot(
    x=top_customers.values,
    y=top_customers.index
)

plt.title("Top 10 Customers by Sales", fontsize=14, fontweight='bold')
plt.xlabel("Total Sales")
plt.ylabel("Customer Name")

# Add value labels
for i, value in enumerate(top_customers.values):
    ax.text(value + 1000, i, f"{value:,.0f}",
            va='center', fontsize=9)

plt.show()
```

```python
## Observation

- The chart highlights the top 10 customers based on total sales.
- A small group of customers contributes significantly to overall revenue.
- These customers can be targeted with loyalty programs and personalized offers to improve customer retention.
```

```python
top_products = (
    df.groupby('Product Name')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

## Observation

- The chart identifies the top 10 products based on total sales.
- A small number of products contribute significantly to the company's revenue.
- These products should receive priority in inventory management and marketing campaigns.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    x='Discount',
    y='Profit',
    data=df
)

plt.title("Discount vs Profit", fontsize=14, fontweight='bold')
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

## Observation

- Higher discounts are generally associated with lower profits.
- Many loss-making transactions occur when high discounts are offered.
- The company should carefully review its discount strategy to avoid unnecessary losses.

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df[['Sales', 'Quantity', 'Discount', 'Profit']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap", fontsize=14, fontweight='bold')

plt.show()
```

# Key Business Insights

1. The Technology category generated the highest sales and profit, making it the strongest-performing category.

2. Furniture generated high sales but comparatively low profit, indicating possible issues with pricing, discounts, or operating costs.

3. Most customer orders were low-value purchases, while a small number of high-value orders contributed significantly to total revenue.

4. Some transactions resulted in negative profits, suggesting that certain discount strategies or products need further investigation.

5. A small group of customers contributed a significant share of total sales, making customer retention important.

6. Sales varied across regions and sub-categories, helping identify high-performing and low-performing business segments.

7. Higher discounts generally showed a tendency to reduce profitability, emphasizing the need for an optimized discount strategy.

# Business Recommendations

- Focus marketing efforts on Technology products, as they generate both high sales and high profit.

- Investigate the Furniture category to identify the reasons behind its low profitability.

- Reduce excessive discounts on products that consistently generate losses.

- Develop loyalty programs for high-value customers to improve customer retention.

- Increase inventory for top-selling products to avoid stock shortages.

- Improve promotional campaigns in lower-performing regions to increase sales.
